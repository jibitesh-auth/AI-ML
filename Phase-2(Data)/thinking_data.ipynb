```python
#Clean & Structure the data
def clean_data1(data):
    text_to_num = {
        "one" : 1, "two" : 2, "three" : 3, "four": 4, "five" : 5
    }
    cleaned_data = []
    unique_user = set()
    for user in data:
        raw_rating = str(user['rating']).strip().lower()
        #str is needed as when fun called 2 time python again checks from start
        if raw_rating in text_to_num:
            raw_rating = text_to_num[raw_rating]
            
        user["rating"] = raw_rating
        #print(raw_rating)
        
        #print(user)

        #Handle missing values
        raw_age = user.get("age")
        if raw_age == None:
            #user.update({"age":None})
            user["age"] = None

        #deduplication
        if user["name"].strip() in unique_user:
            continue
        unique_user.add(user["name"])
        cleaned_data.append(user)

    return cleaned_data
            
        
```

```python
clean_data1(data)
```

## Assignment-2: Getting meaningful insights from data

- What's avg rating?
- What % of users gave poor rating(<3)

```python
#Get meaningful insights from data
def get_insights(data):
    
    #avg rating
    sum = 0
    for user in data:
        sum+= float(user["rating"])
    print(f" avg rating = {sum/len(data)}")

    #percentage of users with poor rating
    poor_rating = 0
    for user in data:
        if float(user["rating"]) < 3:
            poor_rating +=1

    print(f"% of user with poor rating = {poor_rating / len(data) * 100}")
   
```

```python
get_insights(data)
```

## Assignment 3: Build a product recommendation features(Rule based, not ML)

- If user rating>=4, recommend same brand products.
- If user rating<4, recommend different brand products.

```python
#Build the recommendation feature
def get_recommendations(data):
    recommendations = []
    for user in data:
        cur_recomm = {}
        cur_recomm["name"] = user["name"]
        if float(user["rating"]) >= 4:
            cur_recomm["brand"] = "Apple"
        else:
            cur_recomm["brand"] = "Samsung"
        recommendations.append(cur_recomm)
    return recommendations
            
            
```

```python
get_recommendations(data)
```
