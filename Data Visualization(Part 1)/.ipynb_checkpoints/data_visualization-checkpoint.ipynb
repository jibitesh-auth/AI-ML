```python
#Oscar Winner Movies
oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]

years = [2008,2009,2010,2011,2012]
oscar_revenue = [1005 , 170, 427, 133, 232]   #in $M

plt.plot(years,oscar_revenue)
plt.title("Oscar Movies Revenue in each Year")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")
```

```python
#Non-Oscar Winner Movies
non_oscar_movies = ["Slumdog Millionaire","Avatar","Inception","Hugo","Lincoln"]
non_oscar_revenue = [378, 2788, 829, 185, 275] #in $M

# plt.plot(years,oscar_revenue,"o--r", label="Oscar Movies")
plt.plot(years,oscar_revenue, color="r", marker="o", linestyle="-.", linewidth = 5, label="Oscar Movies")
plt.plot(years,non_oscar_revenue,"<-g", label="Non-Oscar Movies")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")

plt.title("Oscar Movies v/s Non-Oscar Movies Revenue")
plt.legend()
# plt.legend(loc="center")
# plt.legend(loc="lower left")
```

```python
plt.plot(years,oscar_revenue, label="Oscar Movies")
plt.plot(years,non_oscar_revenue, label="Non-Oscar Movies")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")

plt.title("Oscar Movies v/s Non-Oscar Movies Revenue")
plt.legend()
# plt.style.use("dark_background")
# plt.style.use("ggplot")
plt.style.use("default")
plt.grid()
plt.show()
```

```python
#cartoonish
with plt.xkcd():
    plt.plot(years,oscar_revenue, label="Oscar Movies")
    plt.plot(years,non_oscar_revenue, label="Non-Oscar Movies")
    plt.xlabel("Years")
    plt.ylabel("Revenue (in $M)")
    
    plt.title("Oscar Movies v/s Non-Oscar Movies Revenue")
    plt.legend()
    plt.grid()
    plt.show()
```
