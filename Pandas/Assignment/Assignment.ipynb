```python
df1[["Name","Sex","Age","Fare","Survived"]]
```

## B.

```python
df1[(df1["Sex"] == "female") & (df1["Fare"] > 30)]
```

## C.

```python
df1.groupby("Pclass").agg(
    Survival_Rate = ("Survived",lambda x: x.mean() * 100),
    Average_Fare = ("Fare","mean"),
    Average_Age = ("Age","mean")
    
)
```
