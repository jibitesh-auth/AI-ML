# Series

```python
s = pd.Series([1,2,3,4,5])
print(s)
print(type(s))

#index
print(s[0])
print(s[1])
```

```python
s = pd.Series([23,24,25,26,"abc"], index=["Ankita","Bear","Bob","Charlie",25])
print(s)

print(s["Ankita"])
#print(s[0])  #Error
print(s.index)

#NOTE: To get the dtype as fully str or int the whole list should be homogenous
#otherwise it will give an object
```

```python
#Vectorized Operation
s1 = pd.Series([1,2,3,4,5])
s2 = pd.Series([10,20,30,40,50])
print(s1+s2)

#Mutable(Immutable size)
s1[0] = 100
print(s1)

changed_s1 = s1.drop(0)
print(changed_s1)
```

# DataFrame

```python
info = {
    "Name": ["Ankita","Bob","Eve"],
    "Age": [23,24,25],
    "GPA": [9.5,8.6,7.2]
}
df = pd.DataFrame(info)
print(df)
print(type(df))
print(df.index)
print(df.columns)
```

#### Creation of DataFrame
