# Series

```python
s = pd.Series([1,2,3,4,5])
print(s)
print(type(s))

#index
print(s[0])
print(s[1])
```

```python
s = pd.Series([23,24,25,26,"abc"], index=["Ankita","Bear","Bob","Charlie",25])
print(s)

print(s["Ankita"])
#print(s[0])  #Error
print(s.index)

#NOTE: To get the dtype as fully str or int the whole list should be homogenous
#otherwise it will give an object
```

```python
#Vectorized Operation
s1 = pd.Series([1,2,3,4,5])
s2 = pd.Series([10,20,30,40,50])
print(s1+s2)

#Mutable(Immutable size)
s1[0] = 100
print(s1)

changed_s1 = s1.drop(0)
print(changed_s1)
```

# Filtering Data

```python
df[df['aqi'] > 100] 
```

```python
df[(df['aqi'] > 100) & (df['temperature'] > 30)]
```

```python
df[(df['aqi'] > 100) & (df['temperature'] < 30)]
```

```python
df[(df["aqi"]>100) & (df["temperature"] > 30)] [['city','aqi']]  #Labels are preserved
```

```python
aqi_data = df[(df['aqi'] > 100) & (df["temperature"] > 30)] [["city","aqi"]]
aqi_data.iloc[0]
```

```python
aqi_data.loc[6]
aqi_data.loc[6:7]   #aqi_data.loc[0] ----> error    
```

### Query()- Filtering

```python
df.query("aqi>100")
```

```python
df.query("aqi>100 & temperature > 30") [["city","aqi"]]
```

```python
#Backticks

df1 = pd.DataFrame([["Adam",23],["Bob",23],["Eve",23]], columns=["Name my","Age"], index=[1,2,3])
print(df1)

df1.query("`Name my`=='Adam'")
```

```python
#python vars
aqi_val = 100
df.query("aqi > @aqi_val")


#Query return a copy[ Whenever we want a copy use .copy no matter it returns a view or a copy]
```
