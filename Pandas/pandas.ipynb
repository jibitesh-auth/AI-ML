# Melt & Pivot

```python
df = pd.DataFrame({
    "country" : ["USA", "USA", "India", "India"],
    "year" : [2020,2021,2020, 2021],
    "sales" : [100,120,90,110],
    "profit": [20,25,18,22]
})

df
```

```python
melted_df = df.melt(
    id_vars = ["country","year"],
    value_vars = ["sales", "profit"],
    var_name = "metrics",
    value_name = "value"
)
    
melted_df
```

```python
original = melted_df.pivot(
    index= ["country","year"],
    columns= "metrics",
    values= "value"
)
original
```

# Merging & Joining data

```python
df_customers = pd.DataFrame({
    "customer_id":[1,2,3,4],
    "name":["Adam","Bob","Charlie","Dave"]
})

df_orders = pd.DataFrame({
    "order_id":[101,102,103,104],
    "customer_id": [2,1,4,5],
    "amount":[250,120,300,180]

})
```

```python
df_customers
```

```python
df_orders
```

```python
#INNER JOIN

pd.merge(df_customers,df_orders, on="customer_id") #(only matching value will be shown)
```

```python
#Left Join

pd.merge(df_customers,df_orders,on="customer_id",how="left") #Keeps all the value of left table
```

```python
#Right Join
pd.merge(df_customers,df_orders,on="customer_id",how="right")
```

```python
#outer join
pd.merge(df_customers,df_orders,on="customer_id",how="outer")

#takes all the values from both the table & assign NAN where value is not there
```
