```python
import requests

URL = "https://www.scrapethissite.com/pages/simple/"
res = requests.get(URL)
```

```python
print(res.status_code)
```

```python
#html content as a string
if res.status_code == 200:
    print(res.text)
```

```python
#raw bytes of the response
if res.status_code == 200:
    print(res.content)
```

```python
#meta data
if res.status_code == 200:
    print(res.headers)
```

```python
with open("scraped_data/data1.html", "w", encoding="utf-8") as f:
    f.write(res.text)
```

```python
from bs4 import BeautifulSoup

with open("scraped_data/data1.html",'r') as f:
    html_content = f.read()

soup = BeautifulSoup(html_content,"lxml")
```

```python
all_h3 = soup.find_all("h3")
type(all_h3)
```

```python
for h3 in all_h3:
    print(h3)
```

```python
# get_text(): to get content btw the tags
for h3 in all_h3:
    name = print(h3.get_text(strip=True))
```

```python
for h3 in all_h3:
    print(h3.find_next('div'))
```

```python
for h3 in all_h3:
    population = print(h3.find_next('div').select("span.country-population")[0].get_text(strip=True))
```

```python
for h3 in all_h3:
    print(h3.find_next('div').select_one("span.country-population").get_text(strip="True"))
```

```python
#Storing Data
all_h3 = soup.find_all("h3")

all_countries = []

for h3 in all_h3:
    name = h3.get_text(strip=True)
    population = h3.find_next('div').select("span.country-population")[0].get_text(strip=True)

    all_countries.append([name,population])
```

```python
all_countries
```

```python
import pandas as pd
df = pd.DataFrame(all_countries,columns=["Name","Population"])
```

```python
df
```

```python
df.to_csv("cleaned_data/data.csv", index=False)
```
